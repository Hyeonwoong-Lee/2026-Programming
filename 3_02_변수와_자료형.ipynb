```python
#문자열 자료형

a = '100'
b = 'python'

#문자열 연산

print(a + b) #문자열간의 연결
print(a * 10) #특정 횟수만큼 반복
print(a < b) #문자의 길이 비교
print('1' in a) #특정 문자열 포함하는가
print('1' not in a) #특정 문자열 포함하지 않는가
```
