```python
def op(num1, num2):
  plus = num1 + num2
  minus = num1 - num2
  multiply = num1 * num2
  divide = num1 / num2
  return plus, minus, multiply, divide

a, b = map(int, input('입력:').split())
p, m, mul, d = op(a, b)
print(f"{a} + {b} = {p}")
print(f"{a} - {b} = {m}")
print(f"{a} * {b} = {mul}")
print(f"{a} / {b} = {d}")
```

```python
def odd_even(num):
  if num % 2 == 0:
    return '짝수'
  else:
    return '홀수'

int_num = int(input('정수를 입력하세요:'))
print(odd_even(int_num))
```
